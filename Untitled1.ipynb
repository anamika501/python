```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()

for i in range(4, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
n = int(input("Enter the first number:"))

for i in range(1, n+1):
    for j in range(i):
        print(i * (j+1), end=" ")
    print()
```

```python
square= lambda a: a* a
rectangle = lambda l, b: l * b
triangle = lambda b, h: 0.5 * b * h

print("Area of square:", square(5))
print("Area of rectangle:", rectangle(5, 6))
print("Area of triangle:",  triangle(2,4))

```
